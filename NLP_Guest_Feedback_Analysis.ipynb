```python
# ============================================================
# STEP 4 — MISSING VALUES AND DUPLICATE INSPECTION
# ============================================================

import os
import pandas as pd
from datetime import datetime

file_path = "Hotel_Reviews.csv"
folder = os.path.join("output", "nlp_results")
os.makedirs(folder, exist_ok=True)

missing_counts = {}
duplicate_count = 0
rows_scanned = 0

for chunk in pd.read_csv(file_path, chunksize=50000, low_memory=False):
    rows_scanned += len(chunk)

    for column in chunk.columns:
        missing_counts[column] = (
            missing_counts.get(column, 0) + chunk[column].isna().sum()
        )

    duplicate_count += int(chunk.duplicated().sum())

quality_df = pd.DataFrame([
    {
        "Column": column,
        "Missing_Count": int(count),
        "Missing_Percentage": round(count / rows_scanned * 100, 2)
    }
    for column, count in missing_counts.items()
])

output_file = os.path.join(folder, "step_04_data_quality_report.csv")
quality_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 4 — MISSING VALUES AND DUPLICATE INSPECTION")
print("=" * 70)
print("Rows scanned:", rows_scanned)
print("Duplicate rows within individual chunks:", duplicate_count)
print("\nMissing value report:")
print(quality_df.to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 4 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 7 — REVIEW TEXT PREPROCESSING
# ============================================================

import os
import re
import pandas as pd

folder = os.path.join("output", "nlp_results")
sample_path = os.path.join(folder, "step_05_nlp_sample.csv")
reviews_df = pd.read_csv(sample_path, low_memory=False)

required_columns = [
    "Positive_Review",
    "Negative_Review",
    "Reviewer_Score"
]

missing_columns = [
    c for c in required_columns if c not in reviews_df.columns
]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

for column in ["Positive_Review", "Negative_Review"]:
    reviews_df[column] = reviews_df[column].fillna("").astype(str)

# Remove dataset placeholders for missing comments
reviews_df["Positive_Review"] = reviews_df["Positive_Review"].replace(
    "No Positive", ""
)
reviews_df["Negative_Review"] = reviews_df["Negative_Review"].replace(
    "No Negative", ""
)

reviews_df["Review_Text"] = (
    reviews_df["Positive_Review"].str.strip()
    + " "
    + reviews_df["Negative_Review"].str.strip()
).str.strip()

def clean_review(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+", " ", text)
    text = re.sub(r"[^a-z\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

reviews_df["Cleaned_Review"] = reviews_df["Review_Text"].apply(clean_review)

before = len(reviews_df)
reviews_df = reviews_df[
    reviews_df["Cleaned_Review"].str.len() > 0
].copy()
removed = before - len(reviews_df)

reviews_df.reset_index(drop=True, inplace=True)

output_file = os.path.join(folder, "step_07_cleaned_reviews.csv")
reviews_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 7 — REVIEW TEXT PREPROCESSING")
print("=" * 70)
print("Rows before removing empty reviews:", before)
print("Empty reviews removed:", removed)
print("Rows after preprocessing:", len(reviews_df))
print("\nReview examples:")
print(reviews_df[["Review_Text", "Cleaned_Review"]].head().to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 7 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 8 — RATING-BASED SENTIMENT LABELS
# ============================================================

import os
import pandas as pd

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(folder, "step_07_cleaned_reviews.csv")
reviews_df = pd.read_csv(input_file, low_memory=False)

reviews_df["Reviewer_Score"] = pd.to_numeric(
    reviews_df["Reviewer_Score"], errors="coerce"
)
reviews_df = reviews_df.dropna(subset=["Reviewer_Score"]).copy()

def score_to_sentiment(score):
    if score >= 7:
        return "Positive"
    elif score >= 5:
        return "Neutral"
    return "Negative"

reviews_df["Rating_Sentiment"] = (
    reviews_df["Reviewer_Score"].apply(score_to_sentiment)
)

distribution = (
    reviews_df["Rating_Sentiment"].value_counts()
    .reindex(["Positive", "Neutral", "Negative"], fill_value=0)
    .rename_axis("Sentiment").reset_index(name="Review_Count")
)

distribution["Percentage"] = (
    distribution["Review_Count"] / len(reviews_df) * 100
).round(2)

output_file = os.path.join(folder, "step_08_rating_sentiment_distribution.csv")
distribution.to_csv(output_file, index=False)

reviews_df.to_csv(input_file, index=False)

print("=" * 70)
print("STEP 8 — RATING-BASED SENTIMENT LABELS")
print("=" * 70)
print(distribution.to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 8 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 10 — VADER SENTIMENT ANALYSIS
# ============================================================

import os
import pandas as pd
from nltk.sentiment import SentimentIntensityAnalyzer

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(folder, "step_07_cleaned_reviews.csv")
reviews_df = pd.read_csv(input_file, low_memory=False)

analyzer = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
    compound = analyzer.polarity_scores(str(text))["compound"]

    if compound >= 0.05:
        sentiment = "Positive"
    elif compound <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return compound, sentiment

scores = reviews_df["Review_Text"].apply(analyze_sentiment)
reviews_df["VADER_Compound"] = scores.apply(lambda x: x[0])
reviews_df["VADER_Sentiment"] = scores.apply(lambda x: x[1])

output_file = os.path.join(folder, "step_10_vader_sentiment_results.csv")
reviews_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 10 — VADER SENTIMENT ANALYSIS")
print("=" * 70)
print("Reviews analyzed:", len(reviews_df))
print("\nSentiment distribution:")
print(reviews_df["VADER_Sentiment"].value_counts().to_string())
print("\nExample results:")
print(reviews_df[
    ["Review_Text", "VADER_Compound", "VADER_Sentiment"]
].head().to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 10 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 13 — RATING AND VADER SENTIMENT COMPARISON
# ============================================================

import os
import pandas as pd
from datetime import datetime

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)

# Recreate rating-based sentiment if needed
if "Rating_Sentiment" not in reviews_df.columns:
    reviews_df["Reviewer_Score"] = pd.to_numeric(
        reviews_df["Reviewer_Score"], errors="coerce"
    )

    def score_to_sentiment(score):
        if score >= 7:
            return "Positive"
        elif score >= 5:
            return "Neutral"
        return "Negative"

    reviews_df["Rating_Sentiment"] = (
        reviews_df["Reviewer_Score"].apply(score_to_sentiment)
    )

# Compare the two methods
comparison = pd.crosstab(
    reviews_df["Rating_Sentiment"],
    reviews_df["VADER_Sentiment"],
    margins=True
)

agreement = (
    reviews_df["Rating_Sentiment"]
    == reviews_df["VADER_Sentiment"]
).mean() * 100

comparison_file = os.path.join(
    folder, "step_13_rating_vader_comparison.csv"
)

comparison.to_csv(comparison_file)

summary = pd.DataFrame([{
    "Total_Reviews": len(reviews_df),
    "Agreement_Percentage": round(agreement, 2),
    "Comparison_Date": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}])

summary_file = os.path.join(
    folder, "step_13_sentiment_agreement_summary.csv"
)
summary.to_csv(summary_file, index=False)

print("=" * 70)
print("STEP 13 — RATING AND VADER SENTIMENT COMPARISON")
print("=" * 70)
print("\nCross-tabulation:")
print(comparison.to_string())
print(f"\nAgreement percentage: {agreement:.2f}%")
print("\nComparison saved to:", os.path.abspath(comparison_file))
print("Agreement summary saved to:", os.path.abspath(summary_file))
print("\nSTEP 13 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 14 — WORD FREQUENCY ANALYSIS
# ============================================================

import os
import re
import pandas as pd
from collections import Counter
from datetime import datetime

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)

# Check required column
if "Cleaned_Review" not in reviews_df.columns:
    raise ValueError("Cleaned_Review column not found.")

# Define common English stop words
stop_words = {
    "the", "and", "was", "for", "with", "that", "this",
    "from", "you", "are", "were", "have", "had", "has",
    "but", "not", "our", "your", "they", "their", "there",
    "very", "all", "can", "would", "could", "been", "its",
    "hotel", "room", "rooms", "stay", "stayed", "one",
    "also", "get", "got", "us", "we", "my", "me"
}

word_counter = Counter()

for text in reviews_df["Cleaned_Review"].fillna("").astype(str):
    words = re.findall(r"\b[a-z']+\b", text.lower())

    meaningful_words = [
        word for word in words
        if len(word) > 2 and word not in stop_words
    ]

    word_counter.update(meaningful_words)

top_words = word_counter.most_common(30)

word_frequency_df = pd.DataFrame(
    top_words,
    columns=["Word", "Frequency"]
)

output_file = os.path.join(
    folder, "step_14_word_frequency_analysis.csv"
)

word_frequency_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 14 — WORD FREQUENCY ANALYSIS")
print("=" * 70)

print("Reviews processed:", len(reviews_df))
print("Unique meaningful words:", len(word_counter))
print("\nTop 30 frequent words:")
print(word_frequency_df.to_string(index=False))

print("\nSaved to:")
print(os.path.abspath(output_file))

print("\nSTEP 14 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 21 — FREQUENT TWO-WORD PHRASE ANALYSIS
# ============================================================

import os
import re
import pandas as pd
from collections import Counter
from datetime import datetime

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)

stop_words = {
    "the", "and", "was", "for", "with", "that", "this",
    "from", "you", "are", "were", "have", "had", "has",
    "but", "not", "our", "your", "they", "their", "there",
    "very", "all", "can", "would", "could", "been", "its",
    "hotel", "room", "rooms", "stay", "stayed", "one",
    "also", "get", "got", "us", "we", "my", "me",
    "when", "which", "out", "too", "only", "little"
}

bigram_counter = Counter()

for text in reviews_df["Cleaned_Review"].fillna("").astype(str):
    words = re.findall(r"\b[a-z]+\b", text.lower())
    words = [
        word for word in words
        if len(word) > 2 and word not in stop_words
    ]

    bigram_counter.update(zip(words, words[1:]))

top_bigrams = bigram_counter.most_common(30)

bigram_df = pd.DataFrame(
    [
        {"Phrase": f"{first} {second}", "Frequency": count}
        for (first, second), count in top_bigrams
    ]
)

output_file = os.path.join(
    folder, "step_21_top_bigrams.csv"
)
bigram_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 21 — FREQUENT TWO-WORD PHRASE ANALYSIS")
print("=" * 70)
print("Reviews processed:", len(reviews_df))
print("\nTop 30 two-word phrases:")
print(bigram_df.to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 21 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 23 — GUEST FEEDBACK THEME ANALYSIS
# ============================================================

import os
import re
import pandas as pd
from datetime import datetime

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)

themes = {
    "Staff": [
        "staff", "employee", "reception", "manager", "friendly"
    ],
    "Cleanliness": [
        "clean", "dirty", "cleanliness", "tidy", "filthy"
    ],
    "Location": [
        "location", "central", "station", "transport", "distance"
    ],
    "Breakfast and Food": [
        "breakfast", "food", "restaurant", "meal", "dinner"
    ],
    "Room Comfort": [
        "bed", "comfortable", "mattress", "pillow", "sleep"
    ],
    "Bathroom": [
        "bathroom", "shower", "toilet", "bath", "water"
    ],
    "Price and Value": [
        "price", "expensive", "cheap", "value", "cost"
    ],
    "Facilities": [
        "wifi", "internet", "lift", "elevator", "facilities"
    ]
}

theme_rows = []

for theme, keywords in themes.items():
    pattern = r"\b(?:" + "|".join(
        re.escape(word) for word in keywords
    ) + r")\b"

    for sentiment in ["Positive", "Neutral", "Negative"]:
        subset = reviews_df[
            reviews_df["VADER_Sentiment"] == sentiment
        ]

        matches = subset["Cleaned_Review"].fillna("").str.contains(
            pattern, regex=True, case=False
        )

        count = int(matches.sum())

        theme_rows.append({
            "Theme": theme,
            "VADER_Sentiment": sentiment,
            "Matching_Reviews": count,
            "Percentage_Within_Sentiment": round(
                count / len(subset) * 100, 2
            ) if len(subset) else 0
        })

theme_df = pd.DataFrame(theme_rows)

output_file = os.path.join(
    folder, "step_23_guest_feedback_themes.csv"
)
theme_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 23 — GUEST FEEDBACK THEME ANALYSIS")
print("=" * 70)
print(theme_df.to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 23 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 26 — SEPARATE POSITIVE AND NEGATIVE COMMENT ANALYSIS
# ============================================================

import os
import pandas as pd
from nltk.sentiment import SentimentIntensityAnalyzer

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)
analyzer = SentimentIntensityAnalyzer()

required_columns = ["Positive_Review", "Negative_Review"]
missing = [c for c in required_columns if c not in reviews_df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

def classify_text(text):
    text = str(text).strip()

    if not text:
        return "Empty"

    score = analyzer.polarity_scores(text)["compound"]

    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    return "Neutral"

reviews_df["Positive_Comment_Sentiment"] = (
    reviews_df["Positive_Review"].fillna("").apply(classify_text)
)

reviews_df["Negative_Comment_Sentiment"] = (
    reviews_df["Negative_Review"].fillna("").apply(classify_text)
)

summary_rows = []

for column, comment_type in [
    ("Positive_Comment_Sentiment", "Positive comment field"),
    ("Negative_Comment_Sentiment", "Negative comment field")
]:
    counts = reviews_df[column].value_counts()

    for sentiment, count in counts.items():
        summary_rows.append({
            "Comment_Field": comment_type,
            "Detected_Sentiment": sentiment,
            "Count": int(count),
            "Percentage": round(count / len(reviews_df) * 100, 2)
        })

summary_df = pd.DataFrame(summary_rows)

output_file = os.path.join(
    folder, "step_26_separate_comment_sentiment.csv"
)
summary_df.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 26 — SEPARATE POSITIVE AND NEGATIVE COMMENT ANALYSIS")
print("=" * 70)
print(summary_df.to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))
print("\nSTEP 26 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 28 — REVIEWER SCORE ANALYSIS BY SENTIMENT
# ============================================================

import os
import pandas as pd
import matplotlib.pyplot as plt

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)

reviews_df["Reviewer_Score"] = pd.to_numeric(
    reviews_df["Reviewer_Score"], errors="coerce"
)

score_summary = (
    reviews_df.dropna(subset=["Reviewer_Score"])
    .groupby("VADER_Sentiment")["Reviewer_Score"]
    .agg(["count", "mean", "median", "std"])
    .round(3)
    .reindex(["Positive", "Neutral", "Negative"])
    .reset_index()
)

output_file = os.path.join(
    folder, "step_28_reviewer_score_analysis.csv"
)
score_summary.to_csv(output_file, index=False)

print("=" * 70)
print("STEP 28 — REVIEWER SCORE ANALYSIS BY SENTIMENT")
print("=" * 70)
print(score_summary.to_string(index=False))
print("\nSaved to:", os.path.abspath(output_file))

plt.figure(figsize=(8, 5))
plt.bar(
    score_summary["VADER_Sentiment"],
    score_summary["mean"]
)
plt.title("Average Reviewer Score by VADER Sentiment")
plt.xlabel("VADER Sentiment")
plt.ylabel("Average Reviewer Score")
plt.tight_layout()

chart_file = os.path.join(
    folder, "step_28_reviewer_score_chart.png"
)
plt.savefig(chart_file, dpi=150, bbox_inches="tight")
plt.show()

print("Chart saved to:", os.path.abspath(chart_file))
print("\nSTEP 28 COMPLETED SUCCESSFULLY")
```

```python
# ============================================================
# STEP 29 — FINAL NLP RESULTS EXPORT
# ============================================================

import os
import pandas as pd
from datetime import datetime

folder = os.path.join("output", "nlp_results")
input_file = os.path.join(
    folder, "step_10_vader_sentiment_results.csv"
)

reviews_df = pd.read_csv(input_file, low_memory=False)

# Add separate comment sentiment results if available
separate_file = os.path.join(
    folder, "step_26_separate_comment_sentiment.csv"
)

if os.path.exists(separate_file):
    from nltk.sentiment import SentimentIntensityAnalyzer
    analyzer = SentimentIntensityAnalyzer()

    def classify_text(text):
        text = str(text).strip()
        if not text:
            return "Empty"
        score = analyzer.polarity_scores(text)["compound"]
        if score >= 0.05:
            return "Positive"
        elif score <= -0.05:
            return "Negative"
        return "Neutral"

    reviews_df["Positive_Comment_Sentiment"] = (
        reviews_df["Positive_Review"].fillna("").apply(classify_text)
    )
    reviews_df["Negative_Comment_Sentiment"] = (
        reviews_df["Negative_Review"].fillna("").apply(classify_text)
    )

# Save final NLP dataset
output_file = os.path.join(
    folder, "step_29_final_nlp_results.csv"
)
reviews_df.to_csv(output_file, index=False)

export_summary = pd.DataFrame([{
    "Rows_Exported": len(reviews_df),
    "Columns_Exported": len(reviews_df.columns),
    "Output_File": os.path.basename(output_file),
    "Export_Date": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}])

summary_file = os.path.join(
    folder, "step_29_final_export_summary.csv"
)
export_summary.to_csv(summary_file, index=False)

print("=" * 70)
print("STEP 29 — FINAL NLP RESULTS EXPORT")
print("=" * 70)
print(export_summary.to_string(index=False))
print("\nDataset saved to:", os.path.abspath(output_file))
print("Export summary saved to:", os.path.abspath(summary_file))
print("\nSTEP 29 COMPLETED SUCCESSFULLY")
```
