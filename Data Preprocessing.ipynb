```python
# ============================================================
# STEP 3 — SET DATASET PATH
# ============================================================

INPUT_FILE = (r"C:\Users\admin\OneDrive\Desktop\Hotel Booking Cancellation Prediction & Analysis\hotel_bookings.csv")
OUTPUT_FOLDER = "output"

PREPROCESSING_FOLDER = os.path.join(
    OUTPUT_FOLDER, "preprocessing_results"
)

EDA_FOLDER = os.path.join(
    OUTPUT_FOLDER, "eda_results"
)

VISUALIZATION_FOLDER = os.path.join(
    OUTPUT_FOLDER, "visualization_results"
)

ML_FOLDER = os.path.join(
    OUTPUT_FOLDER, "machine_learning_results"
)

for folder in [
    OUTPUT_FOLDER,
    PREPROCESSING_FOLDER,
    EDA_FOLDER,
    VISUALIZATION_FOLDER,
    ML_FOLDER
]:
    os.makedirs(folder, exist_ok=True)

print("Dataset path:", INPUT_FILE)
print("Output folders created successfully!")
```

```python
# ============================================================
# STEP 4 — LOAD ORIGINAL DATASET
# ============================================================

print("\n" + "=" * 70)
print("STEP 4 — LOAD ORIGINAL DATASET")
print("=" * 70)

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        "Dataset not found. Please update INPUT_FILE "
        "with the correct hotel_bookings.csv path."
    )

df = pd.read_csv(INPUT_FILE)

print("Dataset loaded successfully!")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nFirst five records:")
print(df.head())
```

```python
# ============================================================
# STEP 5 — BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 5 — BASIC DATASET INFORMATION")
print("=" * 70)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
# ============================================================
# STEP 6 — CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 6 — MISSING VALUE ANALYSIS")
print("=" * 70)

missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_analysis = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage.round(2)
})

missing_analysis = missing_analysis.sort_values(
    "Missing_Count",
    ascending=False
)

print(missing_analysis)

missing_analysis.to_csv(
    os.path.join(
        PREPROCESSING_FOLDER,
        "step_06_missing_value_analysis.csv"
    )
)
```

```python
# ============================================================
# STEP 7 — CHECK DUPLICATE RECORDS
# ============================================================

print("\n" + "=" * 70)
print("STEP 7 — DUPLICATE ANALYSIS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
# ============================================================
# STEP 8 — REMOVE DUPLICATE RECORDS
# ============================================================

print("\n" + "=" * 70)
print("STEP 8 — REMOVE DUPLICATES")
print("=" * 70)

df = df.drop_duplicates().copy()

print("Duplicates removed.")
print("Updated dataset shape:", df.shape)
```

```python
# ============================================================
# STEP 9 — IDENTIFY NUMERICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("STEP 9 — NUMERICAL FEATURES")
print("=" * 70)

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)

print("Number of numerical features:", len(numeric_columns))
```

```python
# ============================================================
# STEP 10 — IDENTIFY CATEGORICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("STEP 10 — CATEGORICAL FEATURES")
print("=" * 70)

categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("Number of categorical features:", len(categorical_columns))
```

```python
# ============================================================
# STEP 11 — HANDLE CHILDREN MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 11 — HANDLE CHILDREN MISSING VALUES")
print("=" * 70)

if "children" in df.columns:
    print("Missing children values before:", df["children"].isnull().sum())

    df["children"] = df["children"].fillna(0)

    print("Missing children values after:", df["children"].isnull().sum())

else:
    print("Children column not found.")
```

```python
# ============================================================
# STEP 12 — HANDLE COUNTRY MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 12 — HANDLE COUNTRY MISSING VALUES")
print("=" * 70)

if "country" in df.columns:
    print("Missing country values before:", df["country"].isnull().sum())

    df["country"] = df["country"].fillna("Unknown")

    print("Missing country values after:", df["country"].isnull().sum())

else:
    print("Country column not found.")
```

```python
# ============================================================
# STEP 13 — HANDLE AGENT MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 13 — HANDLE AGENT MISSING VALUES")
print("=" * 70)

if "agent" in df.columns:
    print("Missing agent values before:", df["agent"].isnull().sum())

    df["agent"] = df["agent"].fillna(0)

    print("Missing agent values after:", df["agent"].isnull().sum())

else:
    print("Agent column not found.")
```

```python
# ============================================================
# STEP 14 — HANDLE COMPANY MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 14 — HANDLE COMPANY MISSING VALUES")
print("=" * 70)

if "company" in df.columns:
    print("Missing company values before:", df["company"].isnull().sum())

    df["company"] = df["company"].fillna(0)

    print("Missing company values after:", df["company"].isnull().sum())

else:
    print("Company column not found.")
```

```python
# ============================================================
# STEP 15 — CREATE TOTAL NIGHTS FEATURE
# ============================================================

print("\n" + "=" * 70)
print("STEP 15 — CREATE TOTAL NIGHTS")
print("=" * 70)

if all(col in df.columns for col in [
    "stays_in_weekend_nights",
    "stays_in_week_nights"
]):
    df["total_nights"] = (
        df["stays_in_weekend_nights"]
        + df["stays_in_week_nights"]
    )

    print("Total nights feature created.")

    print(
        df[
            [
                "stays_in_weekend_nights",
                "stays_in_week_nights",
                "total_nights"
            ]
        ].head()
    )

else:
    print("Required night columns not found.")
```

```python
# ============================================================
# STEP 16 — CREATE TOTAL GUESTS FEATURE
# ============================================================

print("\n" + "=" * 70)
print("STEP 16 — CREATE TOTAL GUESTS")
print("=" * 70)

guest_columns = [
    col for col in ["adults", "children", "babies"]
    if col in df.columns
]

if guest_columns:
    df["total_guests"] = (
        df[guest_columns].fillna(0).sum(axis=1)
    )

    print("Total guests feature created.")
    print(df["total_guests"].describe())

else:
    print("Guest columns not found.")
```

```python
# ============================================================
# STEP 17 — CREATE ARRIVAL DATE
# ============================================================

print("\n" + "=" * 70)
print("STEP 17 — CREATE ARRIVAL DATE")
print("=" * 70)

month_map = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

if "arrival_date_month" in df.columns:
    df["arrival_month_num"] = df[
        "arrival_date_month"
    ].map(month_map)

    if all(col in df.columns for col in [
        "arrival_date_year",
        "arrival_date_day_of_month"
    ]):
        df["arrival_date"] = pd.to_datetime(
            {
                "year": df["arrival_date_year"],
                "month": df["arrival_month_num"],
                "day": df["arrival_date_day_of_month"]
            },
            errors="coerce"
        )

        print("Arrival date feature created.")
        print(df[["arrival_date"]].head())

else:
    print("Arrival month column not found.")
```

```python
# ============================================================
# STEP 18 — CHECK INVALID GUEST RECORDS
# ============================================================

print("\n" + "=" * 70)
print("STEP 18 — INVALID GUEST RECORDS")
print("=" * 70)

if "total_guests" in df.columns:
    zero_guest_count = (df["total_guests"] == 0).sum()

    print("Bookings with zero guests:", zero_guest_count)

    print(
        "These records are reported for review, "
        "not automatically deleted."
    )
# ============================================================
# STEP 19 — OUTLIER ANALYSIS USING IQR
# ============================================================

print("\n" + "=" * 70)
print("STEP 19 — OUTLIER ANALYSIS")
print("=" * 70)

outlier_results = []

for column in df.select_dtypes(include=np.number).columns:

    values = df[column].dropna()

    if values.empty:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (values < lower_bound) |
        (values > upper_bound)
    ).sum()

    outlier_results.append({
        "Feature": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": int(outlier_count)
    })

outlier_df = pd.DataFrame(outlier_results)

print(outlier_df)

outlier_df.to_csv(
    os.path.join(
        PREPROCESSING_FOLDER,
        "step_19_outlier_analysis.csv"
    ),
    index=False
)

print("Outliers identified; not automatically removed.")
# ============================================================
# STEP 20 — SAVE PREPROCESSED DATASET
# ============================================================

print("\n" + "=" * 70)
print("STEP 20 — SAVE PREPROCESSED DATASET")
print("=" * 70)

CLEANED_FILE = os.path.join(
    PREPROCESSING_FOLDER,
    "hotel_bookings_cleaned.csv"
)

df.to_csv(CLEANED_FILE, index=False)

print("Cleaned dataset saved:", CLEANED_FILE)
print("Dataset shape:", df.shape)
# ============================================================
# STEP 21 — FINAL PREPROCESSING SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STEP 21 — PREPROCESSING SUMMARY")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values remaining:", df.isnull().sum().sum())
print("Duplicate records remaining:", df.duplicated().sum())
print("Cleaned file:", CLEANED_FILE)

print("\nMEMBER 2 — PREPROCESSING COMPLETED")


# ============================================================
# STEP 22 — LOAD CLEANED DATASET
# MEMBER 3: EDA & STATISTICAL ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 22 — LOAD CLEANED DATASET")
print("=" * 70)

df = pd.read_csv(CLEANED_FILE)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```
