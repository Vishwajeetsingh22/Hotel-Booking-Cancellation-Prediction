```python
# ============================================================
# MEMBER 3 — EDA & STATISTICAL ANALYSIS
# STEPS 22–56
# ============================================================

# Prerequisite: Run Member 2 first so the cleaned CSV exists.
# This script independently loads the cleaned CSV from the expected output path.
# It also recreates the common output folders if needed.

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid')

INPUT_FILE = (r"C:\Users\admin\OneDrive\Desktop\Hotel Booking Cancellation Prediction & Analysis\output\preprocessing_results\hotel_bookings_cleaned.csv")
OUTPUT_FOLDER = 'output'
PREPROCESSING_FOLDER = os.path.join(OUTPUT_FOLDER, 'preprocessing_results')
EDA_FOLDER = os.path.join(OUTPUT_FOLDER, 'eda_results')
VISUALIZATION_FOLDER = os.path.join(OUTPUT_FOLDER, 'visualization_results')
ML_FOLDER = os.path.join(OUTPUT_FOLDER, 'machine_learning_results')
for folder in [OUTPUT_FOLDER, PREPROCESSING_FOLDER, EDA_FOLDER, VISUALIZATION_FOLDER, ML_FOLDER]:
    os.makedirs(folder, exist_ok=True)

CLEANED_FILE = os.path.join(PREPROCESSING_FOLDER, 'hotel_bookings_cleaned.csv')
if not os.path.exists(CLEANED_FILE):
    raise FileNotFoundError('Run Member 2 first; cleaned dataset not found: ' + CLEANED_FILE)
df = pd.read_csv(CLEANED_FILE)
month_map = {'January': 1, 'February': 2, 'March': 3, 'April': 4, 'May': 5, 'June': 6, 'July': 7, 'August': 8, 'September': 9, 'October': 10, 'November': 11, 'December': 12}
target_column = 'is_canceled'
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(exclude=np.number).columns.tolist()

# ============================================================
# STEP 23 — BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 23 — BASIC DATASET INFORMATION")
print("=" * 70)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
# ============================================================
# STEP 24 — MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 24 — MISSING VALUE ANALYSIS")
print("=" * 70)

missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_analysis = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

print(missing_analysis)

missing_analysis.to_csv(
    os.path.join(
        EDA_FOLDER,
        "step_24_missing_value_analysis.csv"
    )
)
```

```python
# ============================================================
# STEP 25 — DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 25 — DUPLICATE ANALYSIS")
print("=" * 70)

print("Duplicate records:", df.duplicated().sum())
```

```python
# ============================================================
# STEP 26 — IDENTIFY NUMERICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("STEP 26 — NUMERICAL FEATURES")
print("=" * 70)

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical columns:")
print(numeric_columns)
```

```python
# ============================================================
# STEP 27 — IDENTIFY CATEGORICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("STEP 27 — CATEGORICAL FEATURES")
print("=" * 70)

categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)
```

```python
# ============================================================
# STEP 28 — DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("STEP 28 — DESCRIPTIVE STATISTICS")
print("=" * 70)

descriptive_statistics = df.describe().T

print(descriptive_statistics)

descriptive_statistics.to_csv(
    os.path.join(
        EDA_FOLDER,
        "step_28_descriptive_statistics.csv"
    )
)
```

```python
# ============================================================
# STEP 29 — MEAN CALCULATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 29 — MEAN")
print("=" * 70)

mean_values = df[numeric_columns].mean()

print(mean_values)

mean_values.to_csv(
    os.path.join(EDA_FOLDER, "step_29_mean.csv"),
    header=["Mean"]
)
```

```python
# ============================================================
# STEP 30 — MEDIAN CALCULATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 30 — MEDIAN")
print("=" * 70)

median_values = df[numeric_columns].median()

print(median_values)

median_values.to_csv(
    os.path.join(EDA_FOLDER, "step_30_median.csv"),
    header=["Median"]
)
```

```python
# ============================================================
# STEP 31 — MODE CALCULATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 31 — MODE")
print("=" * 70)

mode_values = df[numeric_columns].mode()

print(mode_values)

mode_values.to_csv(
    os.path.join(EDA_FOLDER, "step_31_mode.csv"),
    index=False
)
```

```python
# ============================================================
# STEP 32 — STANDARD DEVIATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 32 — STANDARD DEVIATION")
print("=" * 70)

std_values = df[numeric_columns].std()

print(std_values)

std_values.to_csv(
    os.path.join(EDA_FOLDER, "step_32_standard_deviation.csv"),
    header=["Standard_Deviation"]
)
```

```python
# ============================================================
# STEP 33 — VARIANCE
# ============================================================

print("\n" + "=" * 70)
print("STEP 33 — VARIANCE")
print("=" * 70)

variance_values = df[numeric_columns].var()

print(variance_values)

variance_values.to_csv(
    os.path.join(EDA_FOLDER, "step_33_variance.csv"),
    header=["Variance"]
)
```

```python
# ============================================================
# STEP 34 — MINIMUM VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 34 — MINIMUM VALUES")
print("=" * 70)

minimum_values = df[numeric_columns].min()

print(minimum_values)

minimum_values.to_csv(
    os.path.join(EDA_FOLDER, "step_34_minimum.csv"),
    header=["Minimum"]
)
```

```python
# ============================================================
# STEP 35 — MAXIMUM VALUES
# ============================================================

print("\n" + "=" * 70)
print("STEP 35 — MAXIMUM VALUES")
print("=" * 70)

maximum_values = df[numeric_columns].max()

print(maximum_values)

maximum_values.to_csv(
    os.path.join(EDA_FOLDER, "step_35_maximum.csv"),
    header=["Maximum"]
)
```

```python
# ============================================================
# STEP 36 — RANGE CALCULATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 36 — RANGE")
print("=" * 70)

range_values = (
    df[numeric_columns].max()
    - df[numeric_columns].min()
)

print(range_values)

range_values.to_csv(
    os.path.join(EDA_FOLDER, "step_36_range.csv"),
    header=["Range"]
)
```

```python
# ============================================================
# STEP 37 — QUARTILE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 37 — QUARTILES")
print("=" * 70)

quartiles = df[numeric_columns].quantile([0.25, 0.50, 0.75])

print(quartiles)

quartiles.to_csv(
    os.path.join(EDA_FOLDER, "step_37_quartiles.csv")
)
```

```python
# ============================================================
# STEP 38 — SKEWNESS ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 38 — SKEWNESS")
print("=" * 70)

skewness = df[numeric_columns].skew()

print(skewness.sort_values(ascending=False))

skewness.to_csv(
    os.path.join(EDA_FOLDER, "step_38_skewness.csv"),
    header=["Skewness"]
)
```

```python
# ============================================================
# STEP 39 — KURTOSIS ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 39 — KURTOSIS")
print("=" * 70)

kurtosis = df[numeric_columns].kurtosis()

print(kurtosis.sort_values(ascending=False))

kurtosis.to_csv(
    os.path.join(EDA_FOLDER, "step_39_kurtosis.csv"),
    header=["Kurtosis"]
)
```

```python
# ============================================================
# STEP 40 — DISTRIBUTION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 40 — DISTRIBUTION ANALYSIS")
print("=" * 70)

distribution_summary = pd.DataFrame({
    "Mean": df[numeric_columns].mean(),
    "Median": df[numeric_columns].median(),
    "Std_Dev": df[numeric_columns].std(),
    "Skewness": df[numeric_columns].skew(),
    "Kurtosis": df[numeric_columns].kurtosis()
})

print(distribution_summary)

distribution_summary.to_csv(
    os.path.join(EDA_FOLDER, "step_40_distribution_summary.csv")
)
```

```python
# ============================================================
# STEP 41 — DISTRIBUTION INTERPRETATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 41 — DISTRIBUTION INTERPRETATION")
print("=" * 70)

distribution_interpretation = []

for column in numeric_columns:

    skew_value = df[column].skew()

    if skew_value > 1:
        interpretation = "Highly positively/right skewed"

    elif skew_value > 0.5:
        interpretation = "Moderately positively/right skewed"

    elif skew_value < -1:
        interpretation = "Highly negatively/left skewed"

    elif skew_value < -0.5:
        interpretation = "Moderately negatively/left skewed"

    else:
        interpretation = "Approximately symmetric"

    distribution_interpretation.append({
        "Feature": column,
        "Skewness": skew_value,
        "Interpretation": interpretation
    })

    print(f"{column}: {interpretation}")

distribution_interpretation_df = pd.DataFrame(
    distribution_interpretation
)

distribution_interpretation_df.to_csv(
    os.path.join(
        EDA_FOLDER,
        "step_41_distribution_interpretation.csv"
    ),
    index=False
)
```

```python
# ============================================================
# STEP 42 — UNIQUE VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 42 — UNIQUE VALUE ANALYSIS")
print("=" * 70)

unique_values = df.nunique()

print(unique_values)

unique_values.to_csv(
    os.path.join(EDA_FOLDER, "step_42_unique_values.csv"),
    header=["Unique_Values"]
)
```

```python
# ============================================================
# STEP 43 — CATEGORICAL FREQUENCY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 43 — CATEGORICAL FREQUENCY ANALYSIS")
print("=" * 70)

for column in categorical_columns:

    print("\nColumn:", column)

    frequency = df[column].value_counts(dropna=False)

    print(frequency.head(20))

    safe_name = column.replace(" ", "_").replace("/", "_")

    frequency.to_csv(
        os.path.join(
            EDA_FOLDER,
            f"step_43_{safe_name}_frequency.csv"
        ),
        header=["Frequency"]
    )
```

```python
# ============================================================
# STEP 44 — CANCELLATION TARGET ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 44 — CANCELLATION TARGET ANALYSIS")
print("=" * 70)

target_column = "is_canceled"

if target_column in df.columns:

    target_distribution = df[target_column].value_counts()

    target_percentage = (
        df[target_column].value_counts(normalize=True) * 100
    )

    target_summary = pd.DataFrame({
        "Count": target_distribution,
        "Percentage": target_percentage
    })

    print(target_summary)

    target_summary.to_csv(
        os.path.join(EDA_FOLDER, "step_44_target_distribution.csv")
    )

else:
    raise ValueError("Target column is_canceled not found.")
```

```python
# ============================================================
# STEP 45 — CORRELATION MATRIX
# ============================================================

print("\n" + "=" * 70)
print("STEP 45 — CORRELATION MATRIX")
print("=" * 70)

correlation_matrix = df.select_dtypes(
    include=np.number
).corr()

print(correlation_matrix)

correlation_matrix.to_csv(
    os.path.join(EDA_FOLDER, "step_45_correlation_matrix.csv")
)
```

```python
# ============================================================
# STEP 46 — STRONG CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 46 — STRONG CORRELATIONS")
print("=" * 70)

correlation_pairs = []

for i in range(len(correlation_matrix.columns)):

    for j in range(i + 1, len(correlation_matrix.columns)):

        feature_1 = correlation_matrix.columns[i]
        feature_2 = correlation_matrix.columns[j]

        correlation_value = correlation_matrix.iloc[i, j]

        correlation_pairs.append({
            "Feature_1": feature_1,
            "Feature_2": feature_2,
            "Correlation": correlation_value,
            "Absolute_Correlation": abs(correlation_value)
        })

correlation_pairs_df = pd.DataFrame(correlation_pairs)

correlation_pairs_df = correlation_pairs_df.sort_values(
    "Absolute_Correlation",
    ascending=False
)

print(correlation_pairs_df.head(20))

correlation_pairs_df.to_csv(
    os.path.join(EDA_FOLDER, "step_46_strong_correlations.csv"),
    index=False
)
```

```python
# ============================================================
# STEP 47 — FEATURES CORRELATED WITH CANCELLATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 47 — FEATURES CORRELATED WITH CANCELLATION")
print("=" * 70)

if target_column in correlation_matrix.columns:

    cancellation_correlations = (
        correlation_matrix[target_column]
        .drop(target_column)
        .sort_values(key=abs, ascending=False)
    )

    print(cancellation_correlations)

    cancellation_correlations.to_csv(
        os.path.join(
            EDA_FOLDER,
            "step_47_cancellation_correlations.csv"
        ),
        header=["Correlation"]
    )
```

```python
# ============================================================
# STEP 48 — ANOMALY / OUTLIER DETECTION
# ============================================================

print("\n" + "=" * 70)
print("STEP 48 — ANOMALY DETECTION")
print("=" * 70)

anomaly_summary = []

for column in numeric_columns:

    values = df[column].dropna()

    mean_value = values.mean()
    std_value = values.std()

    if pd.notna(std_value) and std_value != 0:

        z_scores = (values - mean_value) / std_value

        anomaly_count = (abs(z_scores) > 3).sum()

    else:
        anomaly_count = 0

    anomaly_summary.append({
        "Feature": column,
        "Anomalies_ZScore_GT_3": int(anomaly_count)
    })

    print(f"{column}: {anomaly_count} anomalies")

anomaly_df = pd.DataFrame(anomaly_summary)

anomaly_df.to_csv(
    os.path.join(EDA_FOLDER, "step_48_anomaly_analysis.csv"),
    index=False
)
```

```python
# ============================================================
# STEP 49 — FEATURE VARIABILITY ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 49 — FEATURE VARIABILITY")
print("=" * 70)

variability = pd.DataFrame({
    "Mean": df[numeric_columns].mean(),
    "Std_Dev": df[numeric_columns].std(),
    "Variance": df[numeric_columns].var()
})

variability["Coefficient_of_Variation"] = (
    variability["Std_Dev"]
    / variability["Mean"].replace(0, np.nan)
)

variability = variability.sort_values(
    "Coefficient_of_Variation",
    ascending=False
)

print(variability)

variability.to_csv(
    os.path.join(EDA_FOLDER, "step_49_feature_variability.csv")
)
```

```python
# ============================================================
# STEP 50 — MEAN COMPARISON BY CANCELLATION STATUS
# ============================================================

print("\n" + "=" * 70)
print("STEP 50 — MEAN BY CANCELLATION STATUS")
print("=" * 70)

grouped_means = df.groupby(target_column)[numeric_columns].mean().T

print(grouped_means)

grouped_means.to_csv(
    os.path.join(EDA_FOLDER, "step_50_means_by_cancellation.csv")
)
```

```python
# ============================================================
# STEP 51 — MEDIAN COMPARISON BY CANCELLATION STATUS
# ============================================================

print("\n" + "=" * 70)
print("STEP 51 — MEDIAN BY CANCELLATION STATUS")
print("=" * 70)

grouped_medians = df.groupby(target_column)[numeric_columns].median().T

print(grouped_medians)

grouped_medians.to_csv(
    os.path.join(EDA_FOLDER, "step_51_medians_by_cancellation.csv")
)
```

```python
# ============================================================
# STEP 52 — HOTEL TYPE CANCELLATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 52 — HOTEL TYPE ANALYSIS")
print("=" * 70)

if "hotel" in df.columns:

    hotel_analysis = df.groupby("hotel")[target_column].agg(
        Bookings="count",
        Cancellation_Rate="mean"
    )

    hotel_analysis["Cancellation_Percentage"] = (
        hotel_analysis["Cancellation_Rate"] * 100
    )

    print(hotel_analysis)

    hotel_analysis.to_csv(
        os.path.join(EDA_FOLDER, "step_52_hotel_analysis.csv")
    )
```

```python
# ============================================================
# STEP 53 — CUSTOMER TYPE CANCELLATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 53 — CUSTOMER TYPE ANALYSIS")
print("=" * 70)

if "customer_type" in df.columns:

    customer_analysis = df.groupby("customer_type")[target_column].agg(
        Bookings="count",
        Cancellation_Rate="mean"
    )

    customer_analysis["Cancellation_Percentage"] = (
        customer_analysis["Cancellation_Rate"] * 100
    )

    print(customer_analysis)

    customer_analysis.to_csv(
        os.path.join(EDA_FOLDER, "step_53_customer_type_analysis.csv")
    )
```

```python
# ============================================================
# STEP 54 — DEPOSIT TYPE CANCELLATION ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 54 — DEPOSIT TYPE ANALYSIS")
print("=" * 70)

if "deposit_type" in df.columns:

    deposit_analysis = df.groupby("deposit_type")[target_column].agg(
        Bookings="count",
        Cancellation_Rate="mean"
    )

    deposit_analysis["Cancellation_Percentage"] = (
        deposit_analysis["Cancellation_Rate"] * 100
    )

    print(deposit_analysis)

    deposit_analysis.to_csv(
        os.path.join(EDA_FOLDER, "step_54_deposit_analysis.csv")
    )
```

```python
# ============================================================
# STEP 55 — LEAD TIME AND ADR ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("STEP 55 — LEAD TIME AND ADR ANALYSIS")
print("=" * 70)

lead_adr_results = {}

for column in ["lead_time", "adr", "total_nights"]:

    if column in df.columns:

        lead_adr_results[column] = df.groupby(target_column)[column].agg(
            ["count", "mean", "median", "std"]
        )

        print(f"\n{column} by cancellation status:")
        print(lead_adr_results[column])

        lead_adr_results[column].to_csv(
            os.path.join(
                EDA_FOLDER,
                f"step_55_{column}_analysis.csv"
            )
        )
```

```python
# ============================================================
# STEP 56 — FINAL EDA SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STEP 56 — FINAL EDA SUMMARY")
print("=" * 70)

eda_summary = {
    "Total_Records": len(df),
    "Total_Features": df.shape[1],
    "Numerical_Features": len(numeric_columns),
    "Categorical_Features": len(categorical_columns),
    "Missing_Values": int(df.isnull().sum().sum()),
    "Duplicate_Records": int(df.duplicated().sum()),
    "Cancelled_Bookings": int((df[target_column] == 1).sum()),
    "Not_Cancelled_Bookings": int((df[target_column] == 0).sum()),
    "Cancellation_Rate_Percentage": round(
        df[target_column].mean() * 100, 2
    )
}

for key, value in eda_summary.items():
    print(f"{key}: {value}")

pd.DataFrame(
    list(eda_summary.items()),
    columns=["Metric", "Value"]
).to_csv(
    os.path.join(EDA_FOLDER, "step_56_final_eda_summary.csv"),
    index=False
)

print("\nMEMBER 3 — STEPS 22–56 COMPLETED")
```
